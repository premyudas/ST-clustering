```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

from constants import datasets
```

```python
df = pd.read_csv("stats_final.csv")
metrics = {
    "ARI": "ARI",
    "F1_Weighted": "F1 Weighted",
    "F1_Score": "F1 Macro",
    "Run time": "Runtime (seconds)",
}
```

```python
# generate box plots
for col, display_name in metrics.items():
    plt.figure(figsize=(9, 6))
    
    # sns.boxplot groups by 'x' and separates models by 'hue'
    sns.boxplot(
        data=df, 
        x="Dataset", 
        y=col, 
        hue="Model", 
        palette="husl"
    )
    
    # Add titles and labels
    plt.title(f"Model Performance Comparison: {display_name}", fontsize=14, pad=15)
    plt.xlabel("Dataset", fontsize=12)
    plt.ylabel(display_name, fontsize=12)
    
    # Improve layout and display/save the plot
    # plt.grid(axis='y', linestyle='--', alpha=0.5)
    plt.tight_layout()
    
    # plt.savefig(f"boxplot_{col_name}.png", dpi=300)
    
    plt.show()
```

```python
titles = ['Adjusted Rand Index (ARI)', 'Weighted F1 Score', 'Macro F1 Score', 'Runtime']

# Set up a clean Seaborn style
sns.set_theme(style="whitegrid")

for dataset_name, dataset in datasets.items():
    sub_df = df[df["Dataset"] == dataset_name]

    # Create a figure with 4 subplots
    fig, axes = plt.subplots(nrows=4, ncols=1, figsize=(14, 5 * len(metrics)))
    # fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(14, 5 * len(metrics) / 2))
    fig.suptitle(
        f"Benchmarking Performance Metrics for {dataset_name} Dataset", 
        fontsize=22, 
        fontweight='bold', 
        y=0.98  # Lifts the title slightly above the top subplot
    )
    
    for idx, metric in enumerate(metrics.keys()):
        ax = axes[idx]
        
        # Generate the scatter plot
        sns.scatterplot(
            data=sub_df, 
            x='Slide', 
            y=metric, 
            hue='Model', 
            style='Model',     # Gives each model a unique shape (circle, square, X, etc.)
            s=120,             # Marker size
            alpha=0.8,         # Slight transparency to handle overlapping points
            ax=ax,
        )
        
        # Formatting the axes and title
        # ax.set_title(f"{titles[idx]} across {dataset_name} Slides", fontsize=16, fontweight='bold', pad=10)
        ax.set_xlabel("Slide ID", fontsize=12, fontweight='bold')
        ax.set_ylabel(titles[idx], fontsize=12, fontweight='bold')
        
        # Rotate x-axis labels if you have a lot of slides
        ax.set_xticklabels(ax.get_xticklabels(), rotation=45, ha='right')
        
        # Move the legend strictly outside the plot so it doesn't cover data points
        ax.legend(title="Model", bbox_to_anchor=(1.02, 1), loc='upper left', frameon=True)
        
        # Optional: adjust y-axis limits if your scores are strictly between 0 and 1
        # ax.set_ylim(0, 1.05) 

    # Adjust layout to prevent overlapping elements
    plt.tight_layout()

    # Save the plot
    plt.savefig(f"figures/{dataset_name}_stats_scatterplot.png", dpi=300, bbox_inches='tight')
    plt.show()
```

```python
# Average ARI, weighted F1, macro F1

averages_df = df.groupby(['Dataset', 'Model'])[[
    'ARI', 
    'F1_Weighted', 
    'F1_Score', 
    'Run time'
]].mean().reset_index()

averages_df.columns = ['Dataset', 'Model', 'Average ARI', 'Average F1 Weighted', 'Average F1 Macro', 'Average Runtime (s)']

# 4. Sort by Dataset (ascending) and then by Avg_ARI (descending) within each dataset
sorted_averages_df = averages_df.sort_values(
    by=['Dataset', 'Average ARI'], 
    ascending=[True, False]
).reset_index(drop=True)

# Display the sorted result
print(sorted_averages_df)
sorted_averages_df.to_csv("stats_avg.csv")
```
